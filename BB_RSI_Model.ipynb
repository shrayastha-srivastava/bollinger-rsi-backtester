```python
import yfinance
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
RKLB = yfinance.download("RKLB", start = "2024-01-01", end = "2026-10-15")
```

```python
df = pd.DataFrame(RKLB)
```

```python
df.columns = df.columns.get_level_values(0)
```

```python
df["Previous_Close"] = df["Close"].shift(1)
```

```python
df["Changes"] = df["Close"] - df["Previous_Close"]
df
```

```python
df["Gains"] = df["Changes"].clip(lower = 0)
df["Loss"] = df["Changes"].clip(upper = 0).abs()
```

```python
df
```

```python
df
```

```python
df["Gain_Sum"] = df["Gains"].rolling(14).sum()
```

```python
df["Loss_Sum"] = df["Loss"].rolling(14).sum()
```

```python
df
```

```python
df["Avg_Gain"] = df["Gain_Sum"]/14
df["Avg_Loss"] = df["Loss_Sum"]/14
df
```

```python
df["RS"] = df["Avg_Gain"]/df["Avg_Loss"]
```

```python
df
```

```python
df["RSI"] = 100- (100/(1+df["RS"]))
df
```

```python
df["RSI"] = 100- (100/(1+df["RS"]))
df
```

```python
std = df["Close"].rolling(14).std()
m_avg = df["Close"].rolling(14).mean()
df["UB"] = m_avg + 2*std
df["LB"] = m_avg - 2*std
df["M_Avg"] = df["Close"].rolling(14).mean()
df
```

```python
#THEORY
#RSI>70, Close>UB: Sell
#RSI<30, Close<LB: Buy
Sell = df.loc[(df["RSI"]>70) & (df["Close"]>df["UB"]) & (df["Close"]>df["Open"]),["Close","Open","RSI","UB","M_Avg"]].copy()
Buy = df.loc[(df["RSI"]<30) & (df["Close"]<df["LB"]) & (df["Close"]<df["Open"]),["Close","Open","RSI","UB","M_Avg"]].copy()
```

```python
Buy
```

```python
open_buys = []
trades = []

for date, row in df.sort_index().iterrows():

    if date in Sell.index:
        for entry_date, entry_price in open_buys:
            trades.append({
                "Entry": entry_date,
                "Exit": date,
                "Entry_Price": entry_price,
                "Exit_Price": row["Close"],
                "Difference": row["Close"] - entry_price
            })

        open_buys = []


    elif date in Buy.index:
        open_buys.append((date, row["Close"]))

trade_log = pd.DataFrame(trades)
trade_log
```
